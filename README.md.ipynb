```python
 README STRUCTURE
1. Project Title & Metadata
Title: Student Performance Prediction System using Linear Regression
Course: B.Sc. Data Science (Capstone Project)
Name: Nayasa
Roll Number: 2601840005

2. Problem Statement
In modern education analytics, predicting student academic success early allows educators to provide targeted interventions. The objective of this project is to build a supervised machine learning model that predicts a student's Performance Index based on key behavioral and academic features, such as daily study hours, previous exam scores, sleep patterns, extracurricular involvement, and practice test papers completed.


3. Dataset Description
The dataset consists of 10,000 student records with zero missing values across 6 core attributes:

Hours Studied: Total number of hours spent studying daily (Numerical)

Previous Scores: Scores obtained in prior examinations (Numerical)

Extracurricular Activities: Participation in extracurriculars (Categorical: Yes/No, encoded as 1/0)

Sleep Hours: Average daily sleep duration (Numerical)

Sample Question Papers Practiced: Number of practice papers solved (Numerical)

Performance Index: Target variable representing overall academic standing (Numerical, ranging from 10 to 100)


4. Methodology & Implementation Workflow
The project followed a strict 5-stage data science lifecycle inside Google Colab:

Data Acquisition & Loading: Loaded the dataset using Python's pandas library and verified data integrity (df.shape, df.info()).

Data Cleaning & Preprocessing: Inspected for duplicates and handled categorical encoding (mapping Yes/No to 1/0).

Exploratory Data Analysis (EDA): Generated correlation matrices and scatter plots using seaborn and matplotlib to discover feature dependencies.

Model Training: Split the dataset into 80% training and 20% testing sets, then trained a Linear Regression model using scikit-learn.

Evaluation & Visualization: Evaluated model performance using Mean Squared Error (MSE) and the $R^2$ (R-squared) score, plotting actual vs. predicted values.


5. Results & Conclusion

Model Performance: The Linear Regression model achieved a high $R^2$ Score, confirming strong predictive accuracy.

Key Insights:
Hours Studied and Previous Scores have the highest positive regression coefficients, indicating they are the primary drivers of a high Performance Index.

Sleep hours and extracurricular activities provide supplementary support to overall student well-being and balance.

Conclusion: The developed model can successfully forecast student academic results, providing a scalable framework that educational institutions can deploy for early student counseling and performance tracking.
```
